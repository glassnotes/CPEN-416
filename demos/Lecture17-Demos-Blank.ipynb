```python
import pennylane as qml
from pennylane import numpy as np
import matplotlib.pyplot as plt
```

## Demo 1: mixed-state simulation

```python
dev = qml.device('default.mixed', wires=1)
  
@qml.qnode(dev)
def mixed_sim():
    qml.QubitDensityMatrix(np.array([[5, 1], [1, 1]])/6, wires=0)
    return qml.state()
```

```python
mixed_sim()
```
