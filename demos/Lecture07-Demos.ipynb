## Demo 1: superdense coding

<img src="fig/superdense.png" width="600px">  

```python
dev = qp.device("default.qubit", wires=2, shots=10)

def create_entangled_state(wires):
    qp.Hadamard(wires[0])
    qp.CNOT([wires[0], wires[1]])

def alice_prepare_bits(b1, b2, wire):
    if b1 == 1:
        qp.PauliZ(wire)
    if b2 == 1: 
        qp.PauliX(wire)

def bell_basis_rotation(wires):
    qp.adjoint(create_entangled_state)(wires)

@qp.qnode(dev)
def superdense_coding(b1, b2):
    create_entangled_state([0, 1])
    alice_prepare_bits(b1, b2, 0)
    bob_meas_result = qp.measure(1, reset=False)
    bell_basis_rotation([0, 1])
    return qp.sample()
```

```python
superdense_coding(0, 1)
```

```python
qp.draw_mpl(superdense_coding)(0, 1)
```
