## Demo 1: superdense coding

<img src="fig/superdense.png" width="600px">  

```python
dev = qml.device("default.qubit", wires=2, shots=1)

def create_entangled_state():
    qml.Hadamard(wires=0)
    qml.CNOT(wires=[0, 1])

def alice_prepare_bits(b1, b2):
    if b1 == 1:
        qml.PauliZ(wires=0)
    if b2 == 1: 
        qml.PauliX(wires=0)

def bell_basis_rotation():
    qml.adjoint(create_entangled_state)()

@qml.qnode(dev)
def superdense_coding():
    create_entangled_state()
    alice_prepare_bits(1, 0)
    bell_basis_rotation()
    return qml.sample()

superdense_coding()
```
