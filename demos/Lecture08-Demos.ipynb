## Demo 1: Deutsch's algorithm

```python
def oracle():
    func = 2
    
    # 1: f(0) = f(1) = 0
    # Do nothing!
    
    # 2: f(0) = f(1) = 1
    if func == 2:
        qp.CNOT(wires=[0, 1])
        qp.PauliX(wires=0)
        qp.CNOT(wires=[0, 1])
        qp.PauliX(wires=0)
        # Or just 
        # qp.PauliX(wires=1)
        
    # 3: f(0) = 0, f(1) = 1
    if func == 3:
        qp.CNOT(wires=[0, 1])
        
    # 4: f(0) = 1, f(1) = 0
    if func == 4:
        qp.PauliX(wires=0)
        qp.CNOT(wires=[0, 1])
        qp.PauliX(wires=0)
```

```python
dev = qp.device("default.qubit", wires=2, shots=1)

@qp.qnode(dev)
def deutsch():
    # Prep qubit 0 in |+>
    qp.Hadamard(0)

    # Prep qubit 1 in |->
    qp.PauliX(1)
    qp.Hadamard(1)

    # Apply oracle
    oracle()

    # Measure in Hadamard basis
    qp.Hadamard(0)
    return qp.sample(wires=0)
```

```python
deutsch()
```
