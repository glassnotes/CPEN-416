## Demo 2: QPE with built-in template

```python
U = np.kron(qml.matrix(qml.T)(wires=0), np.kron(qml.matrix(qml.PauliX)(wires=0), np.eye(2)))
```

```python
eigvals, eigvecs = np.linalg.eig(U)
```

```python
eigvals
```

```python
target_eigvec = eigvecs[:, 4]
```

```python
target_eigvec
```

```python
eigvals[4]
```

```python
np.angle(eigvals[4]) / (2 * np.pi)
```

```python
estimation_wires = list(range(3))
target_wires = list(range(3, 6))
```

```python
dev = qml.device('default.qubit', wires=6, shots=1)
```

```python
@qml.qnode(dev)
def quantum_phase_estimation():
    qml.MottonenStatePreparation(target_eigvec, wires=target_wires)
    qml.QuantumPhaseEstimation(U, target_wires=target_wires, estimation_wires=estimation_wires)
    return qml.sample(wires=estimation_wires)
```

```python
quantum_phase_estimation()
```

```python
print(qml.draw(qml.transforms.decompose(quantum_phase_estimation, max_expansion=1))())
```
